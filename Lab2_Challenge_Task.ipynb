```python
# [EXERCISE 1]
SURNAME = "BACATAN"
SEED_NUM = 3

# 1. Generated Sensor Data mock values
generated_raw_data = [47.2, "ERROR", 15.8, 89.1, "NULL", 7.4, 61.3, 122.5]

valid_readings = []
invalid_count = 0
log = []

# Add starting log messages
log.append("--- EXECUTION LOG STARTED ---")
log.append("Initializing sensor array validation for Engineer " + SURNAME + ".")

# 2. Process and validate data streams
for entry in generated_raw_data:
    try:
        val = float(entry)
        valid_readings.append(val)
        log.append("Validated reading: " + str(val))
    except ValueError:
        invalid_count = invalid_count + 1
        log.append("CRITICAL: Suppressed invalid input artifact '" + str(entry) + "'")

# 3. Classify valid readings using basic if-else
low_count = 0
nominal_count = 0
high_count = 0

for val in valid_readings:
    if val <= 40.0:
        low_count = low_count + 1
    elif val <= 80.0:
        nominal_count = nominal_count + 1
    else:
        high_count = high_count + 1

classification_results = {
    "Low (0-40)": low_count,
    "Nominal (41-80)": nominal_count,
    "High (>80)": high_count
}

# --- DISPLAY ASSESSMENT DATA ---
print("=== ASSESSMENT DATA ===")
print("Generated Sensor Data:", generated_raw_data)
print("Valid/Invalid Results:", str(len(valid_readings)) + " Valid | " + str(invalid_count) + " Invalid")
print("Classification Results:", classification_results)

print("\nExecution Log:")
for entry in log:
    print("  " + entry)

# Calculate mean safely
mean_val = sum(valid_readings) / len(valid_readings)

print("\nFinal Output:")
print("System processing complete. Mean valid payload value: " + str(round(mean_val, 2)) + "°C")
```
